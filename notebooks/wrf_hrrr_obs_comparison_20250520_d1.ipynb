# WRF vs. HRRR vs. observations: reflectivity and brightness temperature

Two three-column comparisons per run/time: column-max radar reflectivity (WRF | HRRR | MRMS observations) and brightness/cloud-top temperature (WRF | HRRR | GOES observations).

Uses `bnf_wrf_eval.plotting.plot_run_refl_comparison` and `plot_run_tb_comparison`, wrappers that take a time plus the run's name and directory, locate the matching wrfout/HRRR/MRMS/GOES files themselves (downloading the MRMS/GOES observation file on demand if it isn't already cached locally, via `auto_download_mrms`/`auto_download_goes`), and label each figure with a big suptitle. See `src/bnf_wrf_eval/{wrf,hrrr,mrms,goes,plotting}.py` for the implementation.

Run this notebook with the project's `.venv` kernel (`uv sync` first if you haven't).

## Pick a run, a directory, and a time

Each wrapper finds `wrfout_d01_<time>` under `run_dir`, the matching `hrrr.tHHz.wrfnatf00.grib2` file, and the closest-in-time MRMS/GOES observation file, on their own.

## Generate the comparison plots

`refl_levels` sets the reflectivity contour intervals; `ctt_levels` sets the brightness-temperature ones. `auto_download_mrms`/`auto_download_goes=True` fetches the observation file (cached under `mrms_data`/`goes_data` after that) if it isn't already there.

```python
import datetime as dt
import numpy as np
from bnf_wrf_eval.plotting import plot_run_refl_comparison, plot_run_tb_comparison
```

```python
run_name = "2.5-km (D1) Thompson"
run_dir = "../satoshi_testruns/20250520lassobnfwrfhrrr3/rund1"
time = dt.datetime(2025, 5, 21, 1, 0, 0)
fig_refl = plot_run_refl_comparison(
    time,
    run_name,
    run_dir,
    hrrr_base_dir="../satoshi_forcing_data/hrrr/hrrrnat_data",
    mrms_dir="../mrms_data",
    output_base_dir="../outputs",
    refl_levels=np.arange(5, 76, 5),
    auto_download_mrms=True,
)
fig_tb = plot_run_tb_comparison(
    time,
    run_name,
    run_dir,
    hrrr_base_dir="../satoshi_forcing_data/hrrr/hrrrnat_data",
    goes_dir="../goes_data",
    output_base_dir="../outputs",
    auto_download_goes=True,
)
```
