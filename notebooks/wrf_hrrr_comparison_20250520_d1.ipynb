# WRF vs. HRRR: max reflectivity and OLR

Generates the first WRF-vs-HRRR sanity-check comparison plot: column-max radar reflectivity and OLR, for one WRF run (`20250917lassobnfwrfhrrr3`) against the matching HRRR native-level analysis hour.

Uses `bnf_wrf_eval.plotting.plot_run_vs_hrrr`, a wrapper that takes a time plus the run's name and directory, locates the matching wrfout/HRRR files itself, and labels the figure with a big suptitle. It forwards to `plot_max_refl_and_olr`, which handles reading both datasets, reprojecting HRRR onto the WRF domain's Lambert Conformal projection, and laying out the four-panel comparison. See `src/bnf_wrf_eval/{wrf,hrrr,plotting}.py` for the implementation.

Run this notebook with the project's `.venv` kernel (`uv sync` first if you haven't).

## Pick a run, a directory, and a time

`plot_run_vs_hrrr` finds `wrfout_d01_<time>` under `run_dir`, and the matching `hrrr.tHHz.wrfnatf00.grib2` analysis file under the default HRRR native-level directory, on its own.

## Generate the comparison plot

`refl_levels` / `olr_levels` set the contourf contour intervals; adjust these to zoom in on a value range.

```python
import datetime as dt
import numpy as np
from bnf_wrf_eval.plotting import plot_run_vs_hrrr
```

```python
run_name = "2.5-km (D1) Thompson"
run_dir = "../satoshi_testruns/20250520lassobnfwrfhrrr3/rund1"
time = dt.datetime(2025, 5, 21, 1, 0, 0)
fig = plot_run_vs_hrrr(
    time,
    run_name,
    run_dir,
    hrrr_base_dir="../satoshi_forcing_data/hrrr/hrrrnat_data",
    output_base_dir="../outputs",
    refl_levels=np.arange(5, 76, 5),
    olr_levels=np.arange(80, 341, 10),
)
```

```python
run_name = "2.5-km (D1) P3 (mp=52)"
run_dir = "../satoshi_testruns/20250520lassobnfwrfhrrr3/rund1-mp52"
time = dt.datetime(2025, 5, 21, 1, 0, 0)
fig = plot_run_vs_hrrr(
    time,
    run_name,
    run_dir,
    hrrr_base_dir="../satoshi_forcing_data/hrrr/hrrrnat_data",
    output_base_dir="../outputs",
    refl_levels=np.arange(5, 76, 5),
    olr_levels=np.arange(80, 341, 10),
)
```
